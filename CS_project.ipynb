```python
# ============================== Q1
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup
import requests


tesla = yf.Ticker("TSLA")
tesla_history = tesla.history(period="1y")
tesla_stock_close = tesla_history["Close"].resample("ME").mean()


tesla_stock_close.head()
```

```python
# ============================== Q2
tesla_income_df = tesla.quarterly_financials.T.reset_index()
tesla_income_df.columns = ["Date"] + list(tesla_income_df.columns[1:])
tesla_revenue_df = tesla_income_df[["Date", "Total Revenue"]].rename(columns={"Total Revenue": "Revenue"})
tesla_revenue_df["Date"] = pd.to_datetime(tesla_revenue_df["Date"])


tesla_revenue_df.head()
tesla_revenue_df.head()
tesla_revenue_df.head()
```

```python
# ============================== Q3
gamestop = yf.Ticker("GME")
gamestop_history = gamestop.history(period="1y")
gamestop_stock_close = gamestop_history["Close"].resample("ME").mean()

gamestop_stock_close.head()
```

```python
# ============================== Q4
gme_income_df = gamestop.quarterly_financials.T.reset_index()
gme_income_df.columns = ["Date"] + list(gme_income_df.columns[1:])
gamestop_revenue_df = gme_income_df[["Date", "Total Revenue"]].rename(columns={"Total Revenue": "Revenue"})
gamestop_revenue_df["Date"] = pd.to_datetime(gamestop_revenue_df["Date"])


gamestop_revenue_df.head()
gamestop_revenue_df.head()
gamestop_revenue_df.head()
```

```python
# ============================== Q5
fig, ax1 = plt.subplots(figsize=(12,6))
ax1.plot(tesla_stock_close.index, tesla_stock_close.values, color='tab:blue', label='Tesla Stock Price')
ax1.set_ylabel("Stock Price (USD)", color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')


ax2 = ax1.twinx()
ax2.bar(tesla_revenue_df["Date"], tesla_revenue_df["Revenue"], width=20, alpha=0.6, color='tab:orange', label='Revenue')
ax2.set_ylabel("Revenue (USD)", color='tab:orange')
ax2.tick_params(axis='y', labelcolor='tab:orange')


plt.title("Tesla Stock Price vs Revenue Dashboard")
fig.tight_layout()
plt.show()
```

```python
# ============================== Q6
fig, ax1 = plt.subplots(figsize=(12,6))
ax1.plot(gamestop_stock_close.index, gamestop_stock_close.values, color='tab:blue', label='GameStop Stock Price')
ax1.set_ylabel("Stock Price (USD)", color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')


ax2 = ax1.twinx()
ax2.bar(gamestop_revenue_df["Date"], gamestop_revenue_df["Revenue"], width=20, alpha=0.6, color='tab:green', label='Revenue')
ax2.set_ylabel("Revenue (USD)", color='tab:green')
ax2.tick_params(axis='y', labelcolor='tab:green')


plt.title("GameStop Stock Price vs Revenue Dashboard")
fig.tight_layout()
plt.show()
```
