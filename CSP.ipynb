```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Q1: Extract Tesla stock data
tesla = yf.Ticker("TSLA")
tesla_data = tesla.history(period="1y")
tesla_data.reset_index(inplace=True)
tesla_data.head()
```

```python
# Q2: Extract Tesla revenue using yfinance
tesla_financials = tesla.financials.T
tesla_revenue = tesla_financials["Total Revenue"].reset_index()
tesla_revenue.columns = ["Date", "Revenue"]
tesla_revenue["Date"] = pd.to_datetime(tesla_revenue["Date"])
tesla_revenue.tail()
```

```python
# Q3: Extract GameStop stock data
gme = yf.Ticker("GME")
gme_data = gme.history(period="1y")
gme_data.reset_index(inplace=True)
gme_data.head()
```

```python
# Q4: Extract GameStop revenue using yfinance
gme_financials = gme.financials.T
gme_revenue = gme_financials["Total Revenue"].reset_index()
gme_revenue.columns = ["Date", "Revenue"]
gme_revenue["Date"] = pd.to_datetime(gme_revenue["Date"])
gme_revenue.tail()
```

```python
# Q5: Define and use make_graph to plot Tesla stock data
def make_graph(data, title):
    plt.figure(figsize=(10, 5))
    plt.plot(data["Date"], data["Close"], color='tab:blue')
    plt.xlabel("Date")
    plt.ylabel("Stock Price (USD)")
    plt.title(title)
    plt.grid(True)
    plt.tight_layout()
    plt.show()

make_graph(tesla_data, "Tesla Stock Price Over Last Year")
```

```python
# Q6: Use make_graph to plot GameStop stock data
make_graph(gme_data, "GameStop Stock Price Over Last Year")
```
